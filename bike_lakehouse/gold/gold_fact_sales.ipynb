# The Transformation logic

```sql
%python
query = """
SELECT
    sa.order_number,
    pr.product_key,
    cu.customer_key,
    sa.order_date,
    sa.ship_date,
    sa.due_date,
    sa.sales_amount,
    sa.quantity,
    sa.price
FROM silver.crm_sales sa
LEFT JOIN workspace.gold.dim_customers cu
    ON sa.customer_id = cu.customer_id
LEFT JOIN workspace.gold.dim_products pr
    ON sa.product_number = pr.product_number;
"""

df = spark.sql(query)
df.limit(10).display()
```

# Writing gold table

```sql
%python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.fact_sales")
```
